```python
A = {1, 2, 3, 4, 5}
B = {2, 3, 5, 7, 8, 9}
union = A|B
intersection = A&B
diff = A-B
sym = A^B
print("Union of two sets: ", union)
print("Intersection of two sets: ", intersection)
print("Difference of two sets: ", diff)
print("Symmetric Difference of two sets: ", sym)
```

```python
A = {1, 2, 3, 4, 5}
B = {2, 3, 5, 7, 8, 9}
union = A.union(B)
intersection = A.intersection(B)
diff = A.difference(B)
sym = A.symmetric_difference(B)
print("Union of two sets: ", union)
print("Intersection of two sets: ", intersection)
print("Difference of two sets: ", diff)
print("Symmetric Difference of two sets: ", sym)
```
