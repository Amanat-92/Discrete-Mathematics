```python
num1 = float(input("Enter First Number: "))
num2 = float(input("Enter Second Number: "))
add = num1+num2
sub = num1-num2
multi = num1*num2
div = num1/num2
int_div = num1//num2
rem = num1%num2
exp = num1**num2

print("Addition of first and second number is ", add)
print("Subtraction of second number from first number is ", sub)
print("Multiplication of first and second number is ", multi)
print("Division of first and second number is ", div)
print("Integer Division of first and second number is ", int_div)
print("Remainder when first number divided by second number: ", rem)
print("Second number to the power of first number: ", exp)

```

```python
S = {2, 3, 4, 5, 6, 7}
T = {2, 3, 6, 10, 12, 14}
print("Union of two sets: ", S.union(T))
#OR print(S|T) | is used for Union 
print("Intersection of two sets: ", S.intersection(T))
#or print(S&T) where & is used for intersection
print("Difference of two sets: ", S-T)  # - sign is used for difference
print("Symmetric Difference of two sets: ", S^T) #For Symmetric difference
# OR print(S.union(T)-S.intersection(T)) (Union - Intersection = Symmetric difference)
```

```python
S = {2, 3, 4, 5, 6, 7}
T = {2, 3, 6, 10, 12, 14}
union = S|T
intersection = S&T
diff = S-T
sym = S^T
print("Union of two sets: ", union)
#OR print(S|T) | is used for Union 
print("Intersection of two sets: ", intersection)
#or print(S&T) where & is used for intersection
print("Difference of two sets: ", diff)  # - sign is used for difference  OR print(S.difference(T))
print("Symmetric Difference of two sets: ", sym) #For Symmetric difference
# OR print(S.union(T)-S.intersection(T)) (Union - Intersection = Symmetric difference)
```
